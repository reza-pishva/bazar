```python
import pandas as pd
import numpy as np

try:
    # خواندن فایل CSV
    df = pd.read_csv('merged_output2.csv')
    print(f"فایل با موفقیت خوانده شد. تعداد سطرها: {len(df)}")
    
    # نمایش ستون‌های موجود برای بررسی
    print(f"\nستون‌های موجود در فایل: {df.columns.tolist()}")
    
    # 1. تبدیل ستون DATE_MILADI به datetime
    if 'DATE_MILADI' in df.columns:
        df['DATE_MILADI'] = pd.to_datetime(df['DATE_MILADI'], errors='coerce')
        print("✓ ستون DATE_MILADI به تاریخ تبدیل شد.")
    else:
        print("✗ خطا: ستون DATE_MILADI در فایل وجود ندارد!")
        exit()
    
    # 2. ایجاد ستون DELTAPI
    months = df['DATE_MILADI'].dt.month
    df['DELTAPI'] = np.where(
        months.isin([10, 11, 12, 1, 2, 3]), 2,
        np.where(months.isin([4, 5, 6, 7, 8, 9]), 4, np.nan)
    )
    print(f"✓ ستون DELTAPI ایجاد شد.")
    
    # 3. ایجاد ستون P_Dec_old
    if 'ebraz' in df.columns:
        df['P_Dec_old'] = df['ebraz'] * 0.98
        print(f"✓ ستون P_Dec_old ایجاد شد.")
    else:
        print("✗ خطا: ستون ebraz در فایل وجود ندارد!")
        exit()
    
    # 4. ایجاد ستون P_test_old = P_Dec_old - DELTAPI
    df['P_test_old'] = df['P_Dec_old'] - df['DELTAPI']
    print(f"✓ ستون P_test_old ایجاد شد.")
    
    # 5. ایجاد ستون P_Act با مقادیر ستون POWER
    if 'POWER' in df.columns:
        df['P_Act'] = df['POWER']
        print(f"✓ ستون P_Act ایجاد شد (مقادیر ستون POWER).")
    else:
        print("✗ هشدار: ستون POWER در فایل وجود ندارد! ستون P_Act با مقادیر NaN ایجاد شد.")
        df['P_Act'] = np.nan
    
    # 6. ایجاد ستون E_TGU = P_Act * 0.98
    df['E_TGU'] = df['P_Act'] * 0.98
    print(f"✓ ستون E_TGU ایجاد شد (P_Act * 0.98).")
    
    # نمایش اطلاعات ستون‌های جدید
    print("\n" + "="*60)
    print("خلاصه ستون‌های ایجاد شده:")
    print("="*60)
    print(f"1. DELTAPI: مقادیر فصلی (2 برای پاییز/زمستان، 4 برای بهار/تابستان)")
    print(f"2. P_Dec_old: ebraz × 0.98")
    print(f"3. P_test_old: P_Dec_old - DELTAPI")
    print(f"4. P_Act: مقادیر ستون POWER")
    print(f"5. E_TGU: P_Act × 0.98")
    
    # نمایش نمونه‌ای از داده‌ها
    print("\n" + "="*60)
    print("نمونه‌ای از 5 سطر اول داده‌ها:")
    print("="*60)
    
    # ستون‌های مهم برای نمایش
    display_cols = ['DATE_MILADI', 'DELTAPI', 'ebraz', 'P_Dec_old', 'P_test_old']
    if 'POWER' in df.columns:
        display_cols.append('POWER')
    display_cols.extend(['P_Act', 'E_TGU'])
    
    print(df[display_cols].head())
    
    # نمایش آمارهای توصیفی
    print("\n" + "="*60)
    print("آمارهای توصیفی ستون‌های جدید:")
    print("="*60)
    new_columns = ['DELTAPI', 'P_Dec_old', 'P_test_old', 'P_Act', 'E_TGU']
    existing_new_cols = [col for col in new_columns if col in df.columns]
    print(df[existing_new_cols].describe())
    
    # نمایش تعداد مقادیر NaN
    print("\n" + "="*60)
    print("تعداد مقادیر NaN در ستون‌های جدید:")
    print("="*60)
    for col in existing_new_cols:
        nan_count = df[col].isna().sum()
        print(f"{col}: {nan_count} مقدار NaN ({nan_count/len(df)*100:.1f}%)")
    
    # 7. ذخیره فایل نهایی
    output_filename = 'merged_output2_with_all_columns.csv'
    df.to_csv(output_filename, index=False)
    print("\n" + "="*60)
    print(f"✓ فایل نهایی با نام '{output_filename}' ذخیره شد.")
    print(f"✓ تعداد کل سطرها: {len(df)}")
    print(f"✓ تعداد کل ستون‌ها: {len(df.columns)}")
    print(f"✓ ستون‌های جدید ایجاد شده:")
    for i, col in enumerate(existing_new_cols, 1):
        print(f"  {i}. {col}")
    
except FileNotFoundError:
    print("✗ خطا: فایل 'merged_output2.csv' یافت نشد!")
    print("مطمئن شوید فایل در مسیر صحیح قرار دارد.")
except Exception as e:
    print(f"✗ خطای غیرمنتظره: {type(e).__name__}: {e}")
```

```python
import pandas as pd
import numpy as np

# خواندن فایل
df = pd.read_csv('merged_output2.csv')

# تبدیل تاریخ
df['DATE_MILADI'] = pd.to_datetime(df['DATE_MILADI'], errors='coerce')

# 1. ستون DELTAPI
months = df['DATE_MILADI'].dt.month
df['DELTAPI'] = np.where(months.isin([10, 11, 12, 1, 2, 3]), 2,
                         np.where(months.isin([4, 5, 6, 7, 8, 9]), 4, np.nan))

# 2. ستون P_Dec_old
df['P_Dec_old'] = df['ebraz'] * 0.98

# 3. ستون P_test_old
df['P_test_old'] = df['P_Dec_old'] - df['DELTAPI']

# 4. ستون P_Act
df['P_Act'] = df.get('POWER', np.nan)

# 5. ستون E_TGU
df['E_TGU'] = df['P_Act'] * 0.98

# 6. ستون Avcap_Min
if 'practical-6' in df.columns:
    # استخراج ماه و روز
    df['month'] = df['DATE_MILADI'].dt.month
    df['day'] = df['DATE_MILADI'].dt.day
    
    # تعیین بازه 15 خرداد (5 ژوئن) تا 15 شهریور (6 سپتامبر)
    def is_in_summer_range(row):
        month = row['month']
        day = row['day']
        
        # 15 خرداد تا پایان خرداد (5-21 ژوئن)
        if month == 6 and day >= 5:
            return True
        # تیر و مرداد کامل (ژوئیه و آگوست)
        elif month in [7, 8]:
            return True
        # 1 شهریور تا 15 شهریور (23 آگوست تا 6 سپتامبر)
        elif month == 9 and day <= 6:
            return True
        else:
            return False
    
    # اعمال شرط
    df['in_summer'] = df.apply(is_in_summer_range, axis=1)
    df['Avcap_Min'] = np.where(df['in_summer'], 
                               df['practical-6'], 
                               df['practical-6'] - 3)
    
    # حذف ستون‌های موقت
    df = df.drop(columns=['month', 'day', 'in_summer'])
else:
    df['Avcap_Min'] = np.nan

# ذخیره فایل
df.to_csv('merged_output2_processed_complete.csv', index=False)

print("پردازش کامل شد! فایل 'merged_output2_processed_complete.csv' ذخیره شد.")
```

```python
import pandas as pd
import numpy as np
from sklearn.ensemble import GradientBoostingRegressor

# ------------------------------------------------------------
# 1) Load data
# ------------------------------------------------------------
df = pd.read_csv("merged_output2.csv", parse_dates=["DATE_MILADI"])

# ------------------------------------------------------------
# 2) Feature Engineering
# ------------------------------------------------------------
df["hour"] = df["HOUR"]
df["dayofweek"] = df["DATE_MILADI"].dt.dayofweek
df["month"] = df["DATE_MILADI"].dt.month
df["year"] = df["DATE_MILADI"].dt.year  # برای خروجی نهایی

# lag features
for l in [24, 48, 72]:
    df[f"lag_{l}"] = df["POWER"].shift(l)

# rolling features
df["roll_mean_24"] = df["POWER"].rolling(24).mean()
df["roll_std_24"] = df["POWER"].rolling(24).std()

df = df.dropna().reset_index(drop=True)

FEATURES = [
    "hour", "dayofweek", "month",
    "DAMA", "ROTOOBAT",
    "lag_24", "lag_48", "lag_72",
    "roll_mean_24", "roll_std_24"
]
X = df[FEATURES]
y = df["POWER"]

# ------------------------------------------------------------
# 3) Train base models
# ------------------------------------------------------------
model_point_1 = GradientBoostingRegressor(
    n_estimators=300, learning_rate=0.05, max_depth=4, random_state=42
)
model_point_2 = GradientBoostingRegressor(
    n_estimators=200, learning_rate=0.08, max_depth=3, random_state=7
)
model_q30 = GradientBoostingRegressor(
    loss="quantile", alpha=0.30,
    n_estimators=300, learning_rate=0.05, max_depth=3, random_state=99
)

model_point_1.fit(X, y)
model_point_2.fit(X, y)
model_q30.fit(X, y)

# ------------------------------------------------------------
# 4) Predictions
# ------------------------------------------------------------
df["P_xgb"] = model_point_1.predict(X)
df["P_cat"] = model_point_2.predict(X)
df["P_q30"] = model_q30.predict(X)

# ------------------------------------------------------------
# 5) Define Conditions
# ------------------------------------------------------------
df["C_peak"] = df["hour"].isin([19, 20, 21, 22])
df["C_summer"] = df["month"].isin([6, 7, 8])
STD_TH = df["roll_std_24"].quantile(0.75)
df["C_vol"] = df["roll_std_24"] > STD_TH
P_MAX = df["POWER"].quantile(0.99)
df["C_near_max"] = df["P_xgb"] > 0.9 * P_MAX

# ------------------------------------------------------------
# 6) Conditional Ensemble Logic
# ------------------------------------------------------------
def conditional_ensemble(row):
    w_q, w_x, w_c = 0.2, 0.5, 0.3
    if row["C_peak"]:
        w_q += 0.3
    if row["C_summer"]:
        w_q += 0.2
    if row["C_vol"]:
        w_c += 0.2
    if row["C_near_max"]:
        w_x -= 0.2
    
    total = w_q + w_x + w_c
    w_q, w_x, w_c = w_q / total, w_x / total, w_c / total
    
    return w_q * row["P_q30"] + w_x * row["P_xgb"] + w_c * row["P_cat"]

df["P_ensemble"] = df.apply(conditional_ensemble, axis=1)

# ------------------------------------------------------------
# 7) Declaration Limits
# ------------------------------------------------------------
P_NOMINAL = df["POWER"].median()
LOWER = 0.94 * P_NOMINAL
UPPER = 1.03 * P_NOMINAL
df["DECLARED"] = df["P_ensemble"].clip(lower=LOWER, upper=UPPER)

# ------------------------------------------------------------
# 8) Final Output - با تاریخ میلادی و شمسی
# ------------------------------------------------------------
FINAL_OUTPUT_COLS = [
    "DATE_MILADI",      # تاریخ میلادی
    "DATE_SHAMSI",      # تاریخ شمسی (لطفاً مطمئن شوید این اسم دقیقاً در فایل اصلی وجود دارد)
    "POWER", "CODE", "UNIT_NO", "DAMA", "ROTOOBAT",
    "12209_G13", "12210_G13", "practical", "ebraz", "importance_factor",
    "year", "month", "dayofweek", "DECLARED"
]

# اگر ستون DATE_SHAMSI با اسم دیگری در فایل اصلی هست (مثلاً DATE یا SHAMSI_DATE و غیره)، اینجا اسم درست رو بگذارید
df_final = df[FINAL_OUTPUT_COLS]

# ذخیره به صورت CSV (در اکسل به راحتی باز می‌شود)
df_final.to_csv("declared_output_conditional_ensemble.csv", index=False)

# اگر مستقیماً فایل اکسل می‌خواهید:
# df_final.to_excel("declared_output_with_dates.xlsx", index=False)

print("✅ فایل خروجی با تاریخ میلادی و شمسی ایجاد شد:")
print("declared_output_with_dates.csv")
print("\nستون‌های خروجی به ترتیب:")
for col in FINAL_OUTPUT_COLS:
    print(f"- {col}")
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from scipy.spatial.distance import cdist

# =========================================================
# 1) تنظیمات
# =========================================================
INPUT_FILE = "merged_output2.csv"
OUTPUT_FILE = "knn_custom_declared_output.xlsx"

TRAIN_START = "2022-01-01"
TRAIN_END   = "2024-01-01"

TEST_START  = "2024-01-01"
TEST_END    = "2026-01-01"

K = 28

# =========================================================
# 2) Load data
# =========================================================
df = pd.read_csv(INPUT_FILE, parse_dates=["DATE_MILADI"])
df = df.sort_values("DATE_MILADI").reset_index(drop=True)

# =========================================================
# 3) اطمینان از عددی بودن ستون‌ها (رفع dtype object)
# =========================================================
numeric_cols = ["POWER", "DAMA", "ROTOOBAT"]
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# =========================================================
# 4) ویژگی‌های زمانی
# =========================================================
df["hour"] = df["HOUR"]
df["dayofweek"] = df["DATE_MILADI"].dt.dayofweek
df["month"] = df["DATE_MILADI"].dt.month
df["year"] = df["DATE_MILADI"].dt.year

# =========================================================
# 5) ساخت lag ها
# =========================================================
df["lag_24"] = df["POWER"].shift(24)
df["lag_48"] = df["POWER"].shift(48)
df["lag_72"] = df["POWER"].shift(72)

features = ["DAMA", "ROTOOBAT", "lag_24"]

# حذف رکوردهای ناقص
df = df.dropna(subset=features + ["POWER"])

# =========================================================
# 6) Train / Test Split زمانی
# =========================================================
train_df = df[
    (df["DATE_MILADI"] >= TRAIN_START) &
    (df["DATE_MILADI"] <= TRAIN_END)
].copy()

test_df = df[
    (df["DATE_MILADI"] >= TEST_START) &
    (df["DATE_MILADI"] <= TEST_END)
].copy()

# =========================================================
# 7) نرمال‌سازی فقط بر اساس Train
# =========================================================
scaler = StandardScaler()
scaler.fit(train_df[features])

train_df[features] = scaler.transform(train_df[features])
test_df[features] = scaler.transform(test_df[features])

# =========================================================
# 8) KNN سفارشی ساعت‌محور + حذف صفر
# =========================================================
declared = []

for _, row in test_df.iterrows():

    h = row["HOUR"]

    # فقط همان ساعت + POWER غیر صفر
    candidates = train_df[
        (train_df["HOUR"] == h) &
        (train_df["POWER"] != 0)
    ]

    # اگر داده معتبر کافی نیست
    if len(candidates) < K:
        declared.append(np.nan)
        continue

    X_train = candidates[features].astype(float).values
    X_test = row[features].astype(float).values.reshape(1, -1)

    distances = cdist(X_test, X_train, metric="euclidean")[0]

    idx = np.argsort(distances)[:K]

    neigh_power = candidates.iloc[idx]["POWER"].values
    neigh_dist = distances[idx]

    weights = 1.0 / (neigh_dist + 1e-6)

    pred = np.sum(weights * neigh_power) / np.sum(weights)
    declared.append(pred)

test_df["DECLARED"] = declared

# =========================================================
# 9) شرط ابراز
# =========================================================
if "ebraz" in test_df.columns:
    test_df.loc[test_df["ebraz"] == 0, "DECLARED"] = 0

# =========================================================
# 10) خروجی نهایی
# =========================================================
output_cols = [
    "HOUR", "DATE_MILADI", "DATE_SHAMSI", "POWER", "CODE", "UNIT_NO",
    "DAMA", "ROTOOBAT", "12209_G13", "12210_G13",
    "practical", "ebraz", "importance_factor",
    "hour", "dayofweek", "month", "year",
    "lag_24", "lag_48", "lag_72",
    "DECLARED"
]

final_df = test_df[output_cols]
final_df.to_excel(OUTPUT_FILE, index=False)

print("✅ KNN سفارشی بازار برق با موفقیت اجرا شد")
print("Train samples:", len(train_df))
print("Test samples :", len(final_df))
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from scipy.spatial.distance import cdist

# =========================================================
# 1) تنظیمات
# =========================================================
INPUT_FILE = "merged_output2.csv"
OUTPUT_FILE = "knn_best_k_declared_output.xlsx"

TRAIN_START = "2022-01-01"
TRAIN_END   = "2024-01-01"

TEST_START  = "2024-01-01"
TEST_END    = "2026-01-01"

K_LIST = list(range(7, 50, 7))   # 7,14,...,49

# =========================================================
# 2) Load data
# =========================================================
df = pd.read_csv(INPUT_FILE, parse_dates=["DATE_MILADI"])
df = df.sort_values("DATE_MILADI").reset_index(drop=True)

# =========================================================
# 3) عددی‌سازی ستون‌ها
# =========================================================
for col in ["POWER", "DAMA", "ROTOOBAT"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# =========================================================
# 4) ویژگی‌های زمانی
# =========================================================
df["hour"] = df["HOUR"]
df["dayofweek"] = df["DATE_MILADI"].dt.dayofweek
df["month"] = df["DATE_MILADI"].dt.month
df["year"] = df["DATE_MILADI"].dt.year

# =========================================================
# 5) lag ها
# =========================================================
df["lag_24"] = df["POWER"].shift(24)
df["lag_48"] = df["POWER"].shift(48)
df["lag_72"] = df["POWER"].shift(72)

features = ["DAMA", "ROTOOBAT", "lag_24"]

df = df.dropna(subset=features + ["POWER"])

# =========================================================
# 6) Train / Test Split
# =========================================================
train_df = df[
    (df["DATE_MILADI"] >= TRAIN_START) &
    (df["DATE_MILADI"] <= TRAIN_END)
].copy()

test_df = df[
    (df["DATE_MILADI"] >= TEST_START) &
    (df["DATE_MILADI"] <= TEST_END)
].copy()

# =========================================================
# 7) Scaling
# =========================================================
scaler = StandardScaler()
scaler.fit(train_df[features])

train_df[features] = scaler.transform(train_df[features])
test_df[features] = scaler.transform(test_df[features])

# =========================================================
# 8) تابع KNN سفارشی
# =========================================================
def knn_predict(test_df, train_df, K):

    preds = []

    for _, row in test_df.iterrows():

        h = row["HOUR"]

        candidates = train_df[
            (train_df["HOUR"] == h) &
            (train_df["POWER"] != 0)
        ]

        if len(candidates) < K:
            preds.append(np.nan)
            continue

        # 🔴 تبدیل قطعی به float64
        X_train = candidates[features].to_numpy(dtype=np.float64)
        X_test = row[features].to_numpy(dtype=np.float64).reshape(1, -1)

        distances = cdist(X_test, X_train, metric="euclidean")[0]

        idx = np.argsort(distances)[:K]

        neigh_power = candidates.iloc[idx]["POWER"].to_numpy(dtype=np.float64)
        neigh_dist = distances[idx]

        weights = 1.0 / (neigh_dist + 1e-6)

        pred = np.sum(weights * neigh_power) / np.sum(weights)
        preds.append(pred)

    return np.array(preds, dtype=np.float64)

# =========================================================
# 9) تست k های مختلف
# =========================================================
results = []

for K in K_LIST:

    test_df["DECLARED_TMP"] = knn_predict(test_df, train_df, K)

    eval_df = test_df.dropna(subset=["DECLARED_TMP", "POWER"]).copy()

    if "ebraz" in eval_df.columns:
        eval_df = eval_df[eval_df["ebraz"] != 0]

    mae = np.mean(np.abs(eval_df["DECLARED_TMP"] - eval_df["POWER"]))

    results.append({"K": K, "MAE": mae})

    print(f"K={K:<2}  MAE={mae:.3f}")

results_df = pd.DataFrame(results)
best_k = results_df.loc[results_df["MAE"].idxmin(), "K"]

print("\n✅ Best K =", best_k)

# =========================================================
# 10) اجرای نهایی با بهترین K
# =========================================================
test_df["DECLARED"] = knn_predict(test_df, train_df, best_k)

if "ebraz" in test_df.columns:
    test_df.loc[test_df["ebraz"] == 0, "DECLARED"] = 0

# =========================================================
# 11) خروجی نهایی
# =========================================================
output_cols = [
    "HOUR", "DATE_MILADI", "DATE_SHAMSI", "POWER", "CODE", "UNIT_NO",
    "DAMA", "ROTOOBAT", "12209_G13", "12210_G13",
    "practical", "ebraz", "importance_factor",
    "hour", "dayofweek", "month", "year",
    "lag_24", "lag_48", "lag_72",
    "DECLARED"
]

test_df[output_cols].to_excel(OUTPUT_FILE, index=False)
results_df.to_excel("knn_k_mae_summary.xlsx", index=False)

print("📊 MAE summary saved")
print("📁 Final output saved")
```
